```python
#3. Which routes have the highest late-delivery rate?
route_df = deliveries_cleaned.merge(
shipments_cleaned[['shipment_id', 'warehouse_id', 'customer_id']],
on='shipment_id'
)


route_df = route_df.merge(
warehouse_cleaned[['warehouse_id', 'city']],
on='warehouse_id'
).rename(columns={'city': 'warehouse_city'})

route_df = route_df.merge(
customers_cleaned[['customer_id', 'city']],
on='customer_id'
).rename(columns={'city': 'customer_city'})


route_df['route'] = (
route_df['warehouse_city'] + ' → ' + route_df['customer_city']
)

route_stats = (
route_df.groupby('route')
.agg(
total_deliveries=('delivery_id', 'count'),
late_deliveries=('delivery_status',
lambda x: (x == 'Delivered Late').sum())
)
.reset_index()
)

route_stats['late_delivery_rate (%)'] = (
route_stats['late_deliveries']
/ route_stats['total_deliveries']
* 100
).round(2)

route_stats.sort_values(
'late_delivery_rate (%)',
ascending=False
).head(10)
```

```python
# Get top 10 routes
top_routes = (
    route_stats.sort_values(
        'late_delivery_rate (%)',
        ascending=False
    )
    .head(10)
)

plt.figure(figsize=(12, 6))

plt.barh(
    top_routes['route'],
    top_routes['late_delivery_rate (%)'],
    color='blue'
)

plt.title('Top 10 Routes with Highest Late Delivery Rate')
plt.xlabel('Late Delivery Rate (%)')
plt.ylabel('Route')


plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```
