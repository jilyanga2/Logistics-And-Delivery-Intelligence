# Shipment Feature Engineering

```python
import pandas as pd
shipments_cleaned = pd.read_csv("cleaned_supplyguard_data/shipments_cleaned.csv")
shipments_cleaned
```

```python
import pandas as pd 
deliveries_cleaned = pd.read_csv("cleaned_supplyguard_data/deliveries_cleaned.csv")
deliveries_cleaned
```

```python
shipments_cleaned
```

```python
deliveries_cleaned
```

```python
delivery_labels_cleaned = pd.read_csv("cleaned_supplyguard_data/delivery_labels_cleaned.csv")
delivery_labels_cleaned
```

```python
shipdeliv = pd.merge(
    shipments_cleaned, 
    deliveries_cleaned,
    on = "shipment_id" ,
    how = "left"

)
```

```python
shipdeliv
```

```python
finalmerge = shipdeliv.merge(delivery_labels_cleaned)
```

```python
finalmerge
```

```python
finalmerge['shipment_weight'] = (
    finalmerge.groupby('shipment_id')['weight_kg']
    .transform('sum')
)
```

```python
finalmerge['shipment_distance'] = (
    finalmerge.groupby('shipment_id')['distance_km']
    .transform('sum')
)
```

```python
finalmerge['delivery_duration'] = (
    finalmerge.groupby('delivery_id')['delivery_duration_hours']
    .transform('sum')
)
```

```python
finalmerge['late_delivery_flag'] = (
    finalmerge.groupby('delivery_id')['is_late']
    .transform('sum')
)
```

```python
finalmerge
```

```python
finalmerge[["shipment_id", "shipment_weight", "shipment_distance", "delivery_duration", "late_delivery_flag"]].drop_duplicates().merge(shipments_cleaned).to_csv("final_data/shipment_final.csv")
```

# Delivery Feature Engineering

```python
#finalmerge.columns.tolist()  #To check the names of the coloums 
```

```python
finalmerge["delivery_date"] = pd.to_datetime(
    finalmerge["delivery_date"] , 
    errors="coerce"
)
```

```python
#Step 1 - Convert delivery day to a proper datetime format 
finalmerge[["delivery_date", "delivery_hour"]].head()
```

```python
#Step 2- delivery weekday  - return the date in text format. 
finalmerge["delivery_weekday"] = finalmerge["delivery_date"].dt.day_name()
```

```python
#Check: if Step 2 works 
finalmerge[["delivery_date" , "delivery_weekday"]].head()
```

```python
#Step 3 - Customer_shipment_count 

finalmerge["customer_shipment_count"]= (
    finalmerge.groupby("customer_id")["shipment_id"]
    .transform("count")
)
```

```python
#Check: Step 3 
finalmerge[["customer_id", "shipment_id", "customer_shipment_count"]].head(10)
```

```python
#Step 4: Custome_average_delivery_time  - round up the delivery time to the next interger. 
#Make use of delivery_duration_hours 

finalmerge["customer_average_delivery_time"] = (
    finalmerge.groupby("customer_id")["delivery_duration_hours"]
    .transform("mean")
)
```

```python
#Check: Step 4 
finalmerge[["customer_id", "delivery_duration_hours", "customer_average_delivery_time"]].head(10)
```

```python
#Step 5: Distance vs Average
#Dataset: delivery_distance_km

#First caculate the average delivery distance for each customer 

finalmerge["customer_average_distance"] = (
    finalmerge.groupby("customer_id")["delivery_distance_km"]
    .transform("mean")
)

#Second caculate the distance vs average 
finalmerge["distance_vs_average"] =(
    finalmerge["delivery_distance_km"] / finalmerge["customer_average_distance"]
)
```

```python
#Check: Step 5
finalmerge[["customer_id", "delivery_distance_km", "customer_average_distance", "distance_vs_average"]].head(10)
```
