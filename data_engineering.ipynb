# Inspecting Customer and Cleaning


```python
import pandas as pd
customers = pd.read_csv("supplyguard_data/customers.csv")
customers
```

```python
customers.info()
```

```python
# Checking for duplicates
customers[customers.duplicated() == True]
```

```python
#Removing duplicates
customers = customers.drop_duplicates()
```

```python
#Changing date from object to datetime
customers['registration_date'] = pd.to_datetime(customers['registration_date'])
```

```python
#Changing the NaT to Nan
customers['monthly_shipping_budget'] = pd.to_numeric(
customers['monthly_shipping_budget'],
errors='coerce'
)
```

```python
#Changing the money from object to float
customers['monthly_shipping_budget'] = pd.to_numeric(customers['monthly_shipping_budget'])
```
