# Inspecting Customer and Cleaning


```python
import pandas as pd
customers = pd.read_csv("supplyguard_data/customers.csv")
customers
```

```python
customers.info()
```

```python
customers.shape
```

```python
customers.describe()
```

```python
# Checking for duplicates
customers[customers.duplicated() == True]
```

```python
#Removing duplicates
customers = customers.drop_duplicates()
```

```python
#Changing date from object to datetime
customers['registration_date'] = pd.to_datetime(customers['registration_date'])
```

```python
customers['registration_date'] = pd.to_datetime(customers['registration_date'])

today = pd.Timestamp.today().normalize()

future_date = customers[customers['registration_date'] > today]
future_date
```

```python
#Changing the NaT to Nan
customers['monthly_shipping_budget'] = pd.to_numeric(
customers['monthly_shipping_budget'],
errors='coerce'
)
```

```python
#Changing the money from object to float
customers['monthly_shipping_budget'] = pd.to_numeric(customers['monthly_shipping_budget'])
```

```python
customers.monthly_shipping_budget
```

```python
customers.columns
```

```python
customers.customer_name.unique()
```

```python
customers.customer_type.unique()
```

```python
customers.province.unique()
```

```python
#Fixing incosistence
customers.province= customers.province.str.strip()
customers.province= customers.province.str.title()
```

```python
customers.city.unique() 
```

```python
customers.registration_date.unique()
```

```python
customers.monthly_shipping_budget.unique()
```

```python
customers['monthly_shipping_budget'].unique
```

```python
#Checking for missing values
customers.isnull().sum()
```

```python
(customers.isna().sum() / len(customers)) * 100
```

```python
customers[customers['monthly_shipping_budget'].isna()]
```

```python
customers['monthly_shipping_budget'].isna().sum()
```

```python
customers[customers['city'].isna()]["city"]
```

```python
customers['city'].isna().sum()
```

```python
#Checking the Pronvices in which province has cities that has missing value 
customers.loc[customers['city'].isna(), "province"].value_counts()
```

```python
customers['city'] = customers['city'].fillna('Unknown')
```

```python
customers.loc[customers['monthly_shipping_budget'].isna(), "city"].value_counts()
```

```python
customers['monthly_shipping_budget'] = customers['monthly_shipping_budget'].fillna(
customers['monthly_shipping_budget'].median())
```

```python
customers.groupby("city")["monthly_shipping_budget"].median()
```

```python
city_counts = customers.groupby(["province", "city"]).size().reset_index(name= "Customer_count")
city_counts
```

```python
customers.to_csv("cleaned_supplyguard_data/customers_cleaned.csv")
```

# Inspecting Deliveries and Cleaning data

```python
deliveries.columns
```

```python
deliveries.delivery_date.unique()
```

```python
deliveries['delivery_date'] = pd.to_datetime(deliveries['delivery_date'])
```

```python
deliveries['delivery_date'] = pd.to_datetime(deliveries['delivery_date'])

today = pd.Timestamp.today().normalize()

future_date = deliveries[deliveries['delivery_date'] > today]
future_date
```

```python
deliveries.delivery_status.unique()
```

```python
deliveries.delivery_status= deliveries.delivery_status.str.strip()
deliveries.delivery_status= deliveries.delivery_status.str.title()
```

```python
deliveries.delivery_duration_hours.unique()
```

```python
deliveries.attempt_number.unique()
```

```python
deliveries.delivery_hour.unique()
```

```python
deliveries.isnull().sum()
```

```python
deliveries.groupby("delivery_status")["fuel_cost_zar"].median()
```

```python
deliveries["fuel_cost_zar"] = deliveries["fuel_cost_zar"].fillna(
deliveries.groupby("delivery_status")["fuel_cost_zar"].transform("median"))
```

```python
deliveries["fuel_cost_zar"].isna().sum()
```

```python
deliveries["fuel_cost_zar"] = deliveries["fuel_cost_zar"].fillna(deliveries["fuel_cost_zar"].median())
```

```python
deliveries["customer_rating"].describe()
deliveries["customer_rating"].value_counts().sort_index()
```

```python
deliveries["customer_rating"] = deliveries["customer_rating"].fillna(deliveries["customer_rating"].median())
```

```python
deliveries.to_csv("cleaned_supplyguard_data/deliveries_cleaned.csv")
```
