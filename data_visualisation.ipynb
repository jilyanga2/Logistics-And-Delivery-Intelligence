```python
import pandas as pd 
customers_cleaned = pd.read_csv("cleaned_supplyguard_data/customers_cleaned.csv")
customers_cleaned 
```

```python
customers_cleaned.info()
```

```python
#How many customers does the company have?
total_customers = customers_cleaned['customer_id'].nunique()
print(f"Total numbers: {total_customers}")
```

```python
#Which province has the most customers? 
province_customers = (
    customers_cleaned.groupby('province')['customer_id'].nunique().sort_values(ascending=False)
)
print(province_customers)
print("Province with the most customers:", province_customers.idxmax())
```

```python
#Which customer segment sends the most shipment?
customer_shipment = (
    customers_cleaned.groupby('customer_type')['customer_id'].nunique().sort_values(ascending=False)
)
print(customer_shipment)
print("The most customer type:", customer_shipment.idxmax())
```
